### Preparação dos dados

*   Carregamento de bibliotecas



```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
```

### Exploração dos dados

* **Índice de Conciliação por Ano (2020-2024):**

```python
dados = {
    'Ano': [2020, 2021, 2022, 2023, 2024],
    'Índice de Conciliação': [11.2, 12.4, 12.0, 12.3, 10.4]
}

df_indice = pd.DataFrame(dados)
print('Índice de Conciliação por Ano')
df_indice
```


*   Formatação e geração de gráfico:





```python


plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(df_indice['Ano'], df_indice['Índice de Conciliação'], marker='o', linestyle='-', color='lightseagreen')

ax.set_title('Índice de Conciliação de 2020 a 2024', fontsize=14, fontweight='bold')
ax.set_xlabel('Ano', fontsize=10, fontweight='bold')
ax.set_ylabel('Índice (%)', fontsize=10, fontweight='bold')
ax.set_xticks(df_indice['Ano'])

for i, txt in enumerate(df_indice['Índice de Conciliação']):
     ax.annotate(f'{txt:.1f}', (df_indice['Ano'][i], df_indice['Índice de Conciliação'][i]), textcoords="offset points", xytext=(0,12), ha='center', fontsize=10, fontweight='bold', color='lightseagreen')

formatter = mticker.FuncFormatter(lambda y, pos: f'{y:.1f}')
ax.yaxis.set_major_formatter(formatter)

ymin = df_indice['Índice de Conciliação'].min()
ymax = df_indice['Índice de Conciliação'].max()
ax.set_ylim(ymin - 0.5, ymax + 0.5)

ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

* **Índice de Conciliação em Fase de Conhecimento e em Fase de Execução (2020-2024):**

```python
fases = {
    'Ano': [2020, 2021, 2022, 2023, 2024],
    'Conhecimento': [17.4, 18.3, 18.2, 18.4, 16.8],
    'Execução': [7.3, 8.9, 8.6, 9.0, 6.6]
}

df_fase = pd.DataFrame(fases)
print('Índice de Conciliação por Fase de Conhecimento/Fase de Execução')
df_fase
```


*   Formatação e geração de gráfico:


```python
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(df_fase['Ano'], df_fase['Conhecimento'], marker='o', linestyle='-', color='#3498db', label='Fase de Conhecimento')
ax.plot(df_fase['Ano'], df_fase['Execução'], marker='o', linestyle='-', color='#e74c3c', label='Fase de Execução')

ax.set_title('Índice de Conciliação por Fase Processual', fontsize=14, fontweight='bold')
ax.set_xlabel('Ano', fontsize=10, fontweight='bold')
ax.set_ylabel('Índice (%)', fontsize=10, fontweight='bold')
ax.set_xticks(df_fase['Ano'])

for i, txt in enumerate(df_fase['Conhecimento']):
     ax.annotate(f'{txt:.1f}', (df_fase['Ano'][i], df_fase['Conhecimento'][i]), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=10, fontweight='bold', color='#3498db')

for i, txt in enumerate(df_fase['Execução']):
     ax.annotate(f'{txt:.1f}', (df_fase['Ano'][i], df_fase['Execução'][i]), textcoords="offset points", xytext=(0,10), ha='center', fontsize=10, fontweight='bold', color='#e74c3c')

formatter = mticker.FuncFormatter(lambda y, pos: f'{y:.1f}')
ax.yaxis.set_major_formatter(formatter)

geral_ymin = df_fase[['Conhecimento', 'Execução']].min().min()
geral_ymax = df_fase[['Conhecimento', 'Execução']].max().max()
ax.set_ylim(geral_ymin - 0.5, geral_ymax + 1.0)

plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

* **Índice de conciliação dos Juizados Especiais em comparativo com os outros**

```python
orgaos = {
    'Ano': [2020, 2021, 2022, 2023, 2024],
    'Juizados Especiais': [15.9, 19.8, 20.9, 18.6, 18.6],
    'Justiça Federal': [6.1, 9.6, 11.5, 12.3, 10.2],
    'Justiça do Trabalho': [20.6, 22.9, 22.0, 20.2, 19.0]
}

df_orgaos = pd.DataFrame(orgaos)
print('Índice de Conciliação nos Juizados Especiais, na Justiça Federal e na Justiça do Trabalho')
df_fase
```


*   Formatação e geração de gráfico:


```python
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(11, 7))

ax.plot(df_orgaos['Ano'], df_orgaos['Juizados Especiais'], marker='o', linestyle='-', color='#3498db', label='Juizados Especiais')
ax.plot(df_orgaos['Ano'], df_orgaos['Justiça Federal'], marker='o', linestyle='-', color='#e74c3c', label='Justiça Federal')
ax.plot(df_orgaos['Ano'], df_orgaos['Justiça do Trabalho'], marker='o', linestyle='-', color='rebeccapurple', label='Justiça do Trabalho')

ax.set_title('Índice de Conciliação nos Juizados Especiais em comparação aos demais âmbitos', fontsize=14, fontweight='bold')
ax.set_xlabel('Ano', fontsize=10, fontweight='bold')
ax.set_ylabel('Índice (%)', fontsize=10, fontweight='bold')
ax.set_xticks(df_orgaos['Ano'])

for i, txt in enumerate(df_orgaos['Juizados Especiais']):
     ax.annotate(f'{txt:.1f}', (df_orgaos['Ano'][i], df_orgaos['Juizados Especiais'][i]), textcoords="offset points", xytext=(0,-15), ha='center', fontsize=10, fontweight='bold', color='#3498db')

for i, txt in enumerate(df_orgaos['Justiça Federal']):
     ax.annotate(f'{txt:.1f}', (df_orgaos['Ano'][i], df_orgaos['Justiça Federal'][i]), textcoords="offset points", xytext=(0,10), ha='center', fontsize=10, fontweight='bold', color='#e74c3c')

for i, txt in enumerate(df_orgaos['Justiça do Trabalho']):
     ax.annotate(f'{txt:.1f}', (df_orgaos['Ano'][i], df_orgaos['Justiça do Trabalho'][i]), textcoords="offset points", xytext=(0,10), ha='center', fontsize=10, fontweight='bold', color='rebeccapurple')

formatter = mticker.FuncFormatter(lambda y, pos: f'{y:.1f}')
ax.yaxis.set_major_formatter(formatter)

geral_ymin = df_orgaos[['Juizados Especiais', 'Justiça Federal', 'Justiça do Trabalho']].min().min()
geral_ymax = df_orgaos[['Juizados Especiais', 'Justiça Federal', 'Justiça do Trabalho']].max().max()
ax.set_ylim(geral_ymin - 0.5, geral_ymax + 1.0)

plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.10), ncol=3, frameon=False)
ax.grid(alpha=0.3)
plt.show()
```

### 📌 Principais Achados:
1. **Evolução Temporal:** O índice global de conciliação apresentou oscilações, partindo de 11,2% em 2020, alcançando picos em 2021 (12,4%) e 2023 (12,3%), e registrando uma retração para 10,4% em 2024.
2. **Fase de Conhecimento vs. Execução:** A fase de conhecimento mantém índices consideravelmente superiores (oscilando em torno de 16,8% a 18,4%), enquanto a fase de execução permanece como um grande gargalo estrutural (variando entre 6,6% e 9,0%), evidenciando a dificuldade de acordos após a sentença fixada.
3. **Comparativo entre Ramos:** Os Juizados Especias destacam-se com taxas expressivas de conciliação (superando 20% em diversos anos), seguidos pela Justiça do Trabalho e pela Justiça Federal.
